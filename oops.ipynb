```python
#constructors
#used to run methods when object is created
class student:
  def __init__(self,name,age):
    self.name=name
    self.age=age
student1=student("Manaswini",19)
print(student1.name)
print(student1.age)
```

```python
#self
#self refers to the current object.
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
student1 = Student("Manaswini", 20)
print(student1.name)
```

```python
#Instance variables
#instance variables belong to particular object
class student:
  def __init__(self,name,age):
    self.name=name
    self.age=age
student1 = student("manaswini",19)
student2 = student("ashwini",20)
print(student1.name)
print(student2.name)
```

```python
#Methods
#methods are functions defined in a class
class student:
  def __init__(self,name):
    self.name=name
  def display(self):
    print("hello",self.name)
student1 = student("Manaswini")
student1.display()
```

```python
class Calculator:

    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b


c = Calculator()

print(c.add(10, 5))
print(c.subtract(10, 5))
```

```python
#Inheritance
#Inheritance allows one class to acquire the properties and methods of another class.
class animal:
  def eat(self):
    print("Eating")
class Dog(animal):
  def bark(self):
    print("Barking")
dog=Dog()
dog.eat()
dog.bark()
```

```python
#Encapsulation
#Encapsulation means combining data and methods inside a class and controlling how the data is accessed.
#public variable
class Student:

    def __init__(self):
        self.name = "Manaswini"
s = Student()
print(s.name)
```

```python
#private variable
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance
account = BankAccount(5000)
print(account.get_balance())
```

```python
#polymorphism
#Many forms
class Dog:

    def sound(self):
        print("Dog says Woof")
class Cat:

    def sound(self):
        print("Cat says Meow")
dog = Dog()
cat = Cat()
dog.sound()
cat.sound()
```
