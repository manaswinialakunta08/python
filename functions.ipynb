```python
#Returning Multiplevalues
def calculate(a,b):
  add=a+b
  sub=a-b
  return add,sub
x,y=calculate(10,5)
print(x)
print(y)
```

```python
#KeywordArguments
def student(name,age):
  print(name)
  print(age)
student("Manaswini",19)
```

```python
#*args
def add(*numbers):
    total = 0

    for n in numbers:
        total = total + n

    return total

print(add(10, 20))
print(add(10, 20, 30))
print(add(10, 20, 30, 40))
```

```python
#**KWargs(dictinary)
def student(**details):
  print(details)
student(name="Manaswini",age=19)
```

```python
#golbal keyword
x = 10

def change():
    global x
    x = 20

change()

print(x)
```

```python
#Function calling another function
def square(n):
    return n * n

def calculate():
    result = square(5)
    print(result)

calculate()
```

```python
def factorial(n):
  if n==0:
    return 1
  else:
    return n*factorial(n-1)
result=factorial(5)
print(result)
```
