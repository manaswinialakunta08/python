```python
# Create and write to the file
with open("student.txt", "w") as file:
    file.write("Name: Manaswini\n")
    file.write("Branch: CSE\n")
    file.write("Marks: 85\n")

# Read the file
with open("student.txt", "r") as file:
    data = file.read()

# Print the data
print(data)
```
